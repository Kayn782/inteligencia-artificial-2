Resumen extendido de IA y Machine Learning
1. Inteligencia Artificial y patrones complejos no lineales

La Inteligencia Artificial (IA) busca desarrollar sistemas capaces de realizar tareas que normalmente requieren inteligencia humana, como reconocer patrones, clasificar información, hacer predicciones o tomar decisiones.

En particular, el Machine Learning (ML) permite que una computadora aprenda patrones a partir de grandes cantidades de datos, sin tener que programar manualmente todas las reglas.

¿Qué significa "patrones complejos no lineales"?

Un patrón lineal puede representarse aproximadamente mediante una línea recta:

Si X aumenta, Y aumenta proporcionalmente.

Pero muchos problemas reales son no lineales. Por ejemplo:

Reconocimiento de imágenes.
Detección de fraude.
Diagnóstico mediante datos.
Reconocimiento de voz.
Predicción del comportamiento de clientes.
Clasificación de diferentes tipos de objetos.

Los modelos de Machine Learning pueden encontrar relaciones complejas entre muchas variables.

Idea clave:
El objetivo no es simplemente memorizar los datos, sino aprender una función o estructura que permita generalizar a datos nuevos.

3. Fuga de datos — Data Leakage

Este es uno de los conceptos más importantes de tus apuntes.

La fuga de datos ocurre cuando información que no debería estar disponible durante el entrenamiento termina influyendo en el modelo.

Esto puede provocar resultados artificialmente buenos.

Ejemplo

Supongamos que queremos predecir si un paciente tiene una enfermedad.

Tenemos:

Edad
Peso
Presión arterial
Resultados de análisis
Diagnóstico final

Si utilizamos accidentalmente información relacionada con el diagnóstico final como variable de entrada, estamos dando al modelo información que en una situación real no tendría.

El modelo podría obtener una precisión altísima, pero sería engañosa.

¿Cómo evitarla?

Una estrategia importante es:

Separar primero los datos → entrenar el procesamiento solo con entrenamiento → transformar validación/test usando lo aprendido.

Por ejemplo:

Datos originales
       ↓
Train / Test
       ↓
Preprocesamiento aprendido SOLO con Train
       ↓
Modelo
       ↓
Evaluación en Test


La fuga de datos puede ocurrir también durante:

Imputación.
Escalamiento.
Selección de características.
Reducción de dimensionalidad.
Selección de variables.
Optimización de hiperparámetros.
4. Imputación → escalamiento → compresión → clasificador

Esta parte de tus apuntes parece describir un pipeline de Machine Learning.

A. Imputación

La imputación consiste en tratar valores faltantes.

Por ejemplo:

Edad	Peso	Ingreso
25	70	2000
30	?	2500
40	80	?

El signo ? representa datos faltantes.

Podemos sustituirlos utilizando:

Media.
Mediana.
Moda.
Métodos más avanzados.

Por ejemplo, para una variable numérica se puede utilizar la mediana.

Importante: la mediana debe calcularse usando el conjunto de entrenamiento, no usando todo el dataset.

B. Escalamiento

Las variables pueden tener escalas muy diferentes.

Ejemplo:

Edad:       18 - 80
Ingresos:   1.000 - 100.000


Una variable podría dominar numéricamente a otra.

El escalamiento transforma las variables para llevarlas a escalas comparables.

Un método común es StandardScaler, que centra los datos alrededor de 0 y utiliza la desviación estándar para escalarlos.

Esto es especialmente importante para algoritmos basados en:

Distancias.
Gradientes.
SVM.
KNN.
Regresión regularizada.
C. Compresión / reducción dimensional

"Comprimir" los datos puede significar reducir la cantidad de características manteniendo la mayor cantidad posible de información relevante.

Una técnica clásica es PCA (Principal Component Analysis).

Ejemplo:

       ↓

Esto puede:

Reducir el costo computacional.

Facilitar la visualización.
Ayudar a trabajar con datasets de alta dimensionalidad.

Pero hay que tener cuidado: PCA también debe ajustarse únicamente con los datos de entrenamiento para evitar fuga de información.

5. El "truco del kernel"

Este concepto normalmente hace referencia al Kernel Trick, utilizado especialmente en algoritmos como las Support Vector Machines (SVM).

El problema es que algunos datos no pueden separarse fácilmente mediante una frontera lineal.

Imagina:

   ● ● ●
  ●     ●

     ○
   ○ ○ ○


Una línea recta podría no ser suficiente para separar los grupos.

El truco del kernel permite trabajar como si los datos hubieran sido transformados a un espacio de mayor dimensión, donde pueden resultar separables.

La idea simplificada es:

Datos originales → transformación implícita → espacio de características → separación

Lo interesante es que el algoritmo puede calcular relaciones en ese espacio sin tener que construir explícitamente toda la transformación.

Kernels comunes
Linear.
Polynomial.
RBF (Radial Basis Function).
Sigmoid.

El kernel RBF es especialmente utilizado cuando la relación entre las variables es compleja y no lineal.

Idea para examen:

El Kernel Trick permite que algoritmos como SVM encuentren fronteras no lineales utilizando funciones kernel, sin calcular explícitamente la transformación a un espacio de mayor dimensión.

6. Bagging

Tu apunte:

"BAGGING reduce usando árboles"

probablemente se refiere a Bagging (Bootstrap Aggregating).

Bagging es una técnica de ensemble learning, es decir, combina múltiples modelos para obtener un modelo más robusto.

La idea es:

                 Dataset
                    ↓
        ┌───────────┼───────────┐
        ↓           ↓           ↓
      Muestra 1   Muestra 2   Muestra 3
        ↓           ↓           ↓
      Árbol 1     Árbol 2     Árbol 3
        ↓           ↓           ↓
        └───────────┼───────────┘
                    ↓
                Combinación
                    ↓
                Predicción


Cada árbol se entrena con una muestra diferente obtenida mediante bootstrap.

Después se combinan las predicciones.

¿Qué problema ayuda a reducir?

Principalmente la varianza.

Un árbol de decisión individual puede cambiar mucho si cambiamos ligeramente los datos de entrenamiento.

Bagging combina muchos árboles para conseguir mayor estabilidad.

7. Sesgo y varianza

Este es otro concepto fundamental.

Sesgo — Bias

El sesgo representa, simplificando, qué tan lejos está el modelo de capturar correctamente la relación real de los datos.

Un modelo con demasiado sesgo suele ser demasiado simple.

Ejemplo:

Modelo demasiado simple
       ↓
No captura la complejidad
       ↓
Underfitting

Varianza — Variance

La varianza representa cuánto cambia el modelo cuando cambia el conjunto de entrenamiento.

Un modelo con demasiada varianza suele adaptarse demasiado a los datos de entrenamiento.

Modelo demasiado complejo
       ↓
Memoriza características del entrenamiento
       ↓
Overfitting


Una forma sencilla de recordarlo:

Problema	Modelo	Resultado
Alto sesgo	Demasiado simple	Underfitting
Alta varianza	Demasiado complejo	Overfitting
Buen equilibrio	Complejidad adecuada	Buena generalización

El objetivo del Machine Learning es conseguir un equilibrio entre sesgo y varianza.

Para la varianza, la idea matemática básica es medir cuánto se dispersan las predicciones alrededor de su valor medio.

8. GridSearchCV vs RandomizedSearchCV

Tu apunte:

"gridsearchcv costosa, randomized más rápida y no tan costosa"

Correcto, pero hay que matizarlo.

Ambos métodos sirven para encontrar hiperparámetros adecuados.

GridSearchCV

GridSearchCV prueba todas las combinaciones especificadas.

Supongamos:

C = [0.1, 1, 10]
gamma = [0.01, 0.1, 1]


Tenemos:

3 × 3 = 9 combinaciones

Si además usamos validación cruzada de 5 particiones:

9 × 5 = 45 entrenamientos

Si tenemos muchos hiperparámetros, el número puede crecer rápidamente.

RandomizedSearchCV

RandomizedSearchCV selecciona aleatoriamente una cantidad determinada de combinaciones.

Por ejemplo:

100 combinaciones posibles
        ↓
RandomizedSearchCV
        ↓
prueba solamente 20


Puede ser mucho más rápido cuando el espacio de búsqueda es enorme.

Comparación
Característica	GridSearchCV	RandomizedSearchCV
Búsqueda	Exhaustiva	Aleatoria
Costo	Puede ser alto	Generalmente menor
Espacio pequeño	Excelente	También sirve
Espacio grande	Puede ser muy costoso	Muy útil
Número de pruebas	Todas las combinaciones	Número definido de pruebas

Idea clave: RandomizedSearchCV no necesariamente encuentra la combinación óptima absoluta, pero puede encontrar configuraciones muy buenas con mucho menor costo computacional.
